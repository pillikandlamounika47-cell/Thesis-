```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
calendar = pd.read_csv(
    'C:/Users/pilli/Downloads/M5_Thesis_Project/data/raw/calendar.csv'
)

sales = pd.read_csv(
    'C:/Users/pilli/Downloads/M5_Thesis_Project/data/raw/sales_train_validation.csv'
)

prices = pd.read_csv(
    'C:/Users/pilli/Downloads/M5_Thesis_Project/data/raw/sell_prices.csv'
)
```

```python
print("Calendar Shape:", calendar.shape)
print("Sales Shape:", sales.shape)
print("Prices Shape:", prices.shape)
```

```python
sales_sample = sales.iloc[:500]
```

```python
sales_long = sales_sample.melt(
    id_vars=[
        'id',
        'item_id',
        'dept_id',
        'cat_id',
        'store_id',
        'state_id'
    ],
    var_name='d',
    value_name='sales'
)
```

```python
sales_long.head()
```

```python
sales_long.shape
```

```python
sales_long = sales_long.merge(
    calendar,
    on='d',
    how='left'
)
```

```python
sales_long = sales_long.merge(
    prices,
    on=['store_id', 'item_id', 'wm_yr_wk'],
    how='left'
)
```

```python
sales_long['date'] = pd.to_datetime(sales_long['date'])
```

```python
sales_long['year'] = sales_long['date'].dt.year
sales_long['month'] = sales_long['date'].dt.month
sales_long['week'] = sales_long['date'].dt.isocalendar().week
sales_long['weekday'] = sales_long['date'].dt.weekday
```

```python
sales_long.head()
```

```python
sales_long.to_csv(
    'C:/Users/pilli/Downloads/M5_Thesis_Project/data/processed/m5_processed.csv',
    index=False
)
```
