```python
import pandas as pd
import numpy as np
```

```python
sales_long = pd.read_csv(
    'C:/Users/pilli/Downloads/M5_Thesis_Project/data/processed/m5_processed.csv',
    low_memory=False
)
```

```python
sales_long['date'] = pd.to_datetime(sales_long['date'])
```

```python
sales_long = sales_long.sort_values(
    ['id', 'date']
)
```

```python
sales_long['lag_7'] = sales_long.groupby(
    'id'
)['sales'].shift(7)

sales_long['lag_28'] = sales_long.groupby(
    'id'
)['sales'].shift(28)
```

```python
sales_long['rolling_mean_7'] = sales_long.groupby(
    'id'
)['sales'].transform(
    lambda x: x.shift(1).rolling(7).mean()
)

sales_long['rolling_mean_28'] = sales_long.groupby(
    'id'
)['sales'].transform(
    lambda x: x.shift(1).rolling(28).mean()
)
```

```python
sales_long['is_weekend'] = sales_long[
    'weekday'
].isin(['Saturday', 'Sunday']).astype(int)
```

```python
sales_long['month_start'] = sales_long[
    'date'
].dt.is_month_start.astype(int)

sales_long['month_end'] = sales_long[
    'date'
].dt.is_month_end.astype(int)
```

```python
sales_long[
    [
        'sales',
        'lag_7',
        'lag_28',
        'rolling_mean_7',
        'rolling_mean_28'
    ]
].head(20)
```

```python
sales_long = sales_long.dropna()
```
