```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
sales_long = pd.read_csv(
    'C:/Users/pilli/Downloads/M5_Thesis_Project/data/processed/m5_processed.csv',
    low_memory=False
)
```

```python
sales_long['date'] = pd.to_datetime(sales_long['date'])
```

```python
sales_long.head()
```

```python
sales_long.shape
```

```python
sales_long.isnull().sum()
```

```python
daily_sales = sales_long.groupby('date')['sales'].sum()

plt.figure(figsize=(14,5))

daily_sales.plot()

plt.title('Total Daily Sales')

plt.xlabel('Date')

plt.ylabel('Sales')

plt.show()
```

```python
store_sales = sales_long.groupby('store_id')['sales'].sum()

plt.figure(figsize=(10,5))

store_sales.plot(kind='bar')

plt.title('Store-Level Sales')

plt.xlabel('Store')

plt.ylabel('Sales')

plt.show()
```

```python
category_sales = sales_long.groupby('cat_id')['sales'].sum()

plt.figure(figsize=(8,5))

category_sales.plot(kind='bar')

plt.title('Category-Level Sales')

plt.xlabel('Category')

plt.ylabel('Sales')

plt.show()
```

```python
monthly_sales = sales_long.groupby('month')['sales'].sum()

plt.figure(figsize=(10,5))

monthly_sales.plot(kind='line')

plt.title('Monthly Sales Trend')

plt.xlabel('Month')

plt.ylabel('Sales')

plt.show()
```
