```python
import pandas as pd
```

```python
data = pd.read_csv(
    r"C:\Users\pilli\Downloads\M5_Thesis_Project\data\processed\feature_engineered_data.csv"
)

print(data.shape)
```

```python
state_sales = (
    data.groupby('state_id')['sales']
    .sum()
    .reset_index()
)

store_sales = (
    data.groupby('store_id')['sales']
    .sum()
    .reset_index()
)

category_sales = (
    data.groupby('cat_id')['sales']
    .sum()
    .reset_index()
)

department_sales = (
    data.groupby('dept_id')['sales']
    .sum()
    .reset_index()
)

item_sales = (
    data.groupby('item_id')['sales']
    .sum()
    .reset_index()
)

print("Hierarchy created successfully")
```

```python
print("Total Sales:")
print(data['sales'].sum())

print("\nState Total:")
print(state_sales['sales'].sum())

print("\nStore Total:")
print(store_sales['sales'].sum())

print("\nCategory Total:")
print(category_sales['sales'].sum())
```

```python
import os

os.makedirs(
    r"C:\Users\pilli\Downloads\M5_Thesis_Project\results\hierarchy",
    exist_ok=True
)

state_sales.to_csv(
    r"C:\Users\pilli\Downloads\M5_Thesis_Project\results\hierarchy\state_sales.csv",
    index=False
)

store_sales.to_csv(
    r"C:\Users\pilli\Downloads\M5_Thesis_Project\results\hierarchy\store_sales.csv",
    index=False
)

category_sales.to_csv(
    r"C:\Users\pilli\Downloads\M5_Thesis_Project\results\hierarchy\category_sales.csv",
    index=False
)

print("Hierarchy files saved")
```
